### الخطوة 3: اختبار التوليد الصوتي محلياً في النوت بوك

```python
from voicelm_engine import VoiceLMTTS
import IPython.display as ipd

engine = VoiceLMTTS()
text = "بِسْمِ اللَّهِ الرَّحْمَنِ الرَّحِيمِ، مَرْحَبًا بِكُمْ فِي كَاجِل."

wav = engine.synthesize(text, speaker="star04", out_wav="sample.wav")
print(f"✅ تم التوليد بنجاح! مدة الصوت: {len(wav)/24000:.2f} ثانية")
ipd.Audio("sample.wav")
```
